```python
import neo4j
from utils_embeddings import load_embedding_model_std, process_node_embeddings, initialize_collection
from qdrant_client import QdrantClient
from utils_qdrant import COLLECTION_NAME_STD
```

```python
NEO4J_URI = "bolt://localhost:7689"
NEO4J_USER = "neo4j"
NEO4J_PASSWORD = "busticket123"

driver = neo4j.GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USER, NEO4J_PASSWORD))
```

```python
# Load embedding model
model = load_embedding_model_std()
```

```python
# Node labels to process
node_labels = ["POI", "AOI"]
# Batch size for processing
batch_size = 100
```

```python
# Connect to Qdrant
client = QdrantClient("localhost", port=6333)
```

```python
# Initialize Qdrant collection
initialize_collection(client, model, COLLECTION_NAME_STD)
```

```python
# Process each node label
try:
    for label in node_labels:
        process_node_embeddings(driver, model, client, label, COLLECTION_NAME_STD, batch_size)
finally:
    driver.close()
```
