```python
from utils_embeddings import load_embedding_model_nomic, process_node_embeddings3, initialize_collection
from utils_qdrant import COLLECTION_NAME_NOMIC
from qdrant_client import QdrantClient
import pandas as pd
```

```python
data = pd.read_csv('exports/aoi_poi_data_geofox.csv')
```

```python
data
```

```python
# Load embedding model
model = load_embedding_model_nomic()
```

```python
# Node labels to process
node_labels = ["POI", "AOI"]
# Batch size for processing
batch_size = 100
```

```python
# Connect to Qdrant
client = QdrantClient("localhost", port=6333)
```

```python
# Initialize Qdrant collection
initialize_collection(client, model, COLLECTION_NAME_NOMIC)
```

```python
# Process each node label

for label in node_labels:
    data_filtered = data[data['node_type'].apply(lambda x: label in str(x))]
    process_node_embeddings3(data_filtered, model, client, label, COLLECTION_NAME_NOMIC, batch_size)
```
