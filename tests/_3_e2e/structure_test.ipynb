```python
import json

route = dict()
try:
    with open("route.json", "r", encoding="utf-8") as json_data:
        route = json.load(json_data)
except Exception as e:
    print(e)

```

```python
import os
print(os.path.isfile("route.json"))
```

```python
route_stations = route.get("scheduleElements")

for part in route_stations:
    line = part.get('line').get('name')
    print(f'Linie: {line}')
    if line == "Fußweg":
        print(f'Startpunkt: {part.get('from').get('name')}')
    else:
        print(f'Haltestelle: {part.get('from').get('name')}')
        print(f'Fahrtrichtung: {part.get('line').get('direction')}')
    print(f'Ziel: {part.get('to').get('name')}')
    print("\n")
```
