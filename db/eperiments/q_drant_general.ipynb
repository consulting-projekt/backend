```python
import sys
sys.path.append("../")
import pandas as pd
import pandas as pd
from utils_embeddings import load_embedding_model_std
from qdrant_client import QdrantClient
from qdrant_client.models import Filter, FieldCondition, GeoRadius, GeoPoint
```

```python
# 1. Verbindungen herstellen
client = QdrantClient("localhost", port=6333)
```

```python
# 1. Get a list of all collections
collections_response = client.get_collections()

# The response contains a list of CollectionDescription objects
collections = collections_response.collections

print(f"Found {len(collections)} collections:")

# 2. Iterate through each collection to get its entry count
for collection in collections:
    collection_name = collection.name
    
    # Get detailed information about the collection, which includes the points count
    collection_info = client.get_collection(collection_name=collection_name)
    points_count = collection_info.points_count
    
    print(f"- Collection '{collection_name}' has {points_count} entries.")
```

```python
def quick_delete_all_collections(client):
    """Delete all collections without confirmation"""
    collections = client.get_collections()
    collection_names = [collection.name for collection in collections.collections]
    
    for name in collection_names:
        client.delete_collection(collection_name=name)
        print(f"Deleted collection: {name}")
    
    print(f"Successfully deleted {len(collection_names)} collections.")
```
