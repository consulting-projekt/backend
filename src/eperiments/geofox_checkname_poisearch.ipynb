```python
def check_name(client, sdName,  maxList=5):
    endpoint = 'checkName' 
    
    request = {
    "language": "de",
    "version": 59,
    "tariffDetails": True,
    "maxList": maxList,
    "theName": sdName,
    "allowTypeSwitch": True
    }

    res = client.send(endpoint, request)

    return res
```
