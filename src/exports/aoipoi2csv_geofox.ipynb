```python
import neo4j
import pandas as pd
```

```python
NEO4J_URI = "bolt://localhost:7689"
NEO4J_USER = "neo4j"
NEO4J_PASSWORD = "busticket123"

driver = neo4j.GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USER, NEO4J_PASSWORD))
```

```python
query = """
MATCH (n)
WHERE (n:AOI OR n:POI)
  AND n.geofoxid IS NOT NULL
RETURN 
    n.name AS name,
    n.description AS description,
    n.tags AS tags,
    labels(n) AS node_type,
    CASE 
        WHEN 'POI' IN labels(n) THEN n.location 
        ELSE null 
    END AS location,
    CASE 
        WHEN 'AOI' IN labels(n) THEN n.centroid 
        ELSE null 
    END AS centroid
"""
```

```python
# Execute query and process results
with driver.session() as session:
    result = session.run(query)
    records = result.data()
```

```python
# Convert to DataFrame
df = pd.DataFrame(records)
```

```python
# Generate filename with current timestamp
filename = f"aoi_poi_data_geofox.csv"

# Save to CSV
df.to_csv(filename, index=False)

print(f"Data successfully exported to {filename}")

# Close the driver connection
driver.close()
```
